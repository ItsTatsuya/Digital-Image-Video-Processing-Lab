```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load the input image
image_path = '../../input.jpg'
original_image = cv2.imread(image_path)

# Check if image loaded successfully
if original_image is None:
    print("Error: Could not load image from", image_path)
else:
    print("Image loaded successfully. Shape:", original_image.shape)
```

```python
# Apply Laplacian filter for sharpening
# Convert to float for computation
original_float = original_image.astype(np.float32)

# Compute Laplacian
laplacian = cv2.Laplacian(original_float, cv2.CV_32F)

# Sharpened image: original + Laplacian (with scaling factor)
alpha = 1.0  # Sharpening factor
sharpened_float = original_float + alpha * laplacian

# Clip to [0, 255] and convert back to uint8
sharpened_image = np.clip(sharpened_float, 0, 255).astype(np.uint8)

print("Sharpening applied with Laplacian filter, alpha =", alpha)
```

```python
# Convert BGR to RGB for display
original_rgb = cv2.cvtColor(original_image, cv2.COLOR_BGR2RGB)
sharpened_rgb = cv2.cvtColor(sharpened_image, cv2.COLOR_BGR2RGB)

# Display the images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(original_rgb)
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(sharpened_rgb)
plt.title('Sharpened Image (Laplacian)')
plt.axis('off')

plt.show()
```

```python
# Save the sharpened image
output_path = 'sharpened_image.jpg'
cv2.imwrite(output_path, sharpened_image)
print("Sharpened image saved as:", output_path)
```
